```python
# We are going to predict salary for employee for repective number of years in a company which is only 2 years old.
# We have salary data for 2 years in place
```

```python
# Pandas is like Excel inside Python — it makes reading, cleaning, and preparing your data for ML very easy.
# Without Pandas, handling data would be much harder using plain Python lists or dictionaries.

import pandas as pd
```

```python
# pd -Short alias for pandas (because we wrote import pandas as pd)
# .read_csv()- A pandas function that reads a CSV file
# "salary_data.csv" - The name of your file you want to load
# dataset - A variable name where the loaded data is stored

dataset=pd.read_csv("Salary_Data.csv")
```

```python
dataset
```

```python
# dataset - Your full loaded Data (all columns)
# [["YearsExperience"]] - Selecting only the YearsExperience column
# independent - Variable where the selected column is stored
independent=dataset[["YearsExperience"]]
```

```python
#In Machine Learning, we have two types of variables
# Independent - Feature / Input / X : The value we give to the model
# Dependent - Target / Output / y : The value we want to predict
```

```python
dependent=dataset[["Salary"]]
```

#### Sklearn is a complete Machine Learning toolkit — it gives you ready-made algorithms, data splitting, evaluation, and preprocessing tools so you can build ML models quickly and easily without coding them from scratch.

```python
# sklearnShort for Scikit-learn — a popular Python ML library
# model_selection - A module inside sklearn for splitting & selecting data
# train_test_split - The specific function we are importing

from sklearn.model_selection import train_test_split
train_test_split(independent, dependent, test_size=0.3, random_state=0)
```

```python
x_train, x_test, y_train, y_test=train_test_split(independent, dependent, test_size=0.3, random_state=0)
```

```python
# sklearn - Scikit-learn ML library
# linear_model - Module inside sklearn for regression models
# LinearRegression - The specific algorithm we are importing
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(x_train, y_train)
```

```python
# Linear Regression finds the best straight line that fits your data
Salary (y)
    |                        * (predicted)
    |                   *
    |              *
    |         *
    |    *
    |________________________
        1    2    3    4     Years Experience (x)
# LinearRegression() - Creates a blank/empty model — not trained yet
# regressor - Variable that stores the model
```

### What happens inside .fit()?
Step 1: Model looks at x_train (YearsExperience)
        and y_train (Salary) together

Step 2: It tries different lines and finds the
        BEST line that fits the data

Step 3: It calculates and stores:
        → m (slope)
        → c (intercept)

Step 4: Model is now TRAINED and ready to predict!

```python
# regressorYour trained Linear Regression model
# .coef_Attribute that gives the slope (m) of the line
# slopeVariable that stores the slope value
slope=regressor.coef_
slope
```

```python
# regressor - Your trained Linear Regression model
# .intercept_ - Attribute that gives the bias/intercept (c) of the line
# bias - Variable that stores the intercept value
bias=regressor.intercept_
bias
```

Salary = slope × YearsExperience + bias (Y=MX+C)

Salary = 9360.26 × YearsExperience + 26777

```python
# regressor - Your trained Linear Regression model
# .predict()- Function that predicts output using learned formula
# x_test- 30% test data (YearsExperience) — unseen by modely_predStores the predicted salaries
y_pred=regressor.predict(x_test)
```

```python
# sklearn.metrics - Module inside sklearn for model evaluation
# r2_score - Function to calculate R² (R-Squared) value
 
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
```

#### Pickle is a Python library used to save and load Python objects (like your trained ML model) into a file.
Think of it like freezing your trained model into a file so you can reuse it later without retraining!

```python
import pickle
filename="Finalized_model_linear.sav" 
# filename - Variable storing the name of the save file
# "Finalized_model_linear.sav"The file where model will be saved
# .savFile extension meaning saved model file
pickle.dump(regressor,open(filename,'wb'))
# open() - Python function to open/create a file
# filename - The file name "Finalized_model_linear.sav"
# 'wb'Mode — write + binary (ML models need this binary format)
```

```python
loaded_model=pickle.load(open(filename,'rb'))
# pickle.load() - Function that reads and restores the saved model
# loaded_model - Variable that stores the restored model
```

### Saving vs Loading
#### Saving - pickle.dump(regressor,open(filename,'wb')) - Freeze model into a file
#### Loading - pickle.load(open(filename,'rb')) - Restore model from file

```python
result=loaded_model.predict([[15]])
```

```python
# loaded_model - Restored trained model
# .predict() - Function to predict salary
# [[15]] - Input employee with 15 years of experience
# result - stores the predicted salary
```

Why [[15]] with Double Brackets?
ML models expect 2D input:
loaded_model.predict([15])    # ❌ 1D - Wrong!
loaded_model.predict([[15]])  # ✅ 2D - Correct!
